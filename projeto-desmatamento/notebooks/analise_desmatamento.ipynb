# Análise de Desmatamento no Brasil

**1. Introdução e Contextualização ambiental:** 
O desmatamento é um dos principais desafios ambientais do Brasil, impactando a biodiversidade e as mudanças climáticas. Este relatório explora esses dados.

**2. Explicação da base:** 
Utilizamos a base `simulacao_desmatamento_brasil.csv`, contendo métricas de áreas desmatadas, preservadas e queimadas entre 2015 e 2024.


```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('../dados/simulacao_desmatamento_brasil.csv')

df.head()
```

```python
area_total = df['area_desmatada_km2'].sum()
focos_queimada = df['focos_queimada'].sum()

print(f"Área Total Desmatada: {area_total:.0f} km²")
print(f"Total de Focos de Queimada: {focos_queimada:.0f}")
```

```python
df_bioma = df.groupby('bioma')['area_desmatada_km2'].sum().reset_index()
fig = px.bar(df_bioma, x='bioma', y='area_desmatada_km2', title='Desmatamento por Bioma')
fig.show()
```

**6. Interpretação e Conclusão:** 
Por meio da análise exploratória, é possível identificar os biomas mais pressionados pelo desmatamento. Os KPIs e visualizações indicam uma forte necessidade de políticas de contenção e fiscalização direcionadas às regiões com maiores picos de desmatamento e queimadas, garantindo assim a conservação das áreas preservadas restantes.
